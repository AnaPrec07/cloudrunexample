```python
import subprocess
import requests

# Get your personal identity token from gcloud
def get_personal_token() -> str:
    result = subprocess.run(
        ["gcloud", "auth", "print-identity-token"],
        capture_output=True,
        text=True,
        check=True
    )
    return result.stdout.strip()

# ── Config ────────────────────────────────────────────────────────────────────
SERVICE_URL = "https://cloudrun-agent-516153587458.us-central1.run.app"  # your actual URL
# SERVICE_URL = "http://0.0.0.0:8080"

# ── Get token ─────────────────────────────────────────────────────────────────
token = get_personal_token()

# ── Call /invoke ──────────────────────────────────────────────────────────────
response = requests.post(
    f"{SERVICE_URL}/invoke",
    headers={
        "Authorization": f"Bearer {token}",
        "Content-Type": "application/json"
    },
    json={
        "user_id": "u1",
        "message": "hello"
    }
)


print(response.status_code)
print(response.json())
```
